## Lab 3 (Część 2): Funkcje (Programowanie proceduralne)

```python
# Definiowanie prostej funkcji (zauważ użycie 'return')
def celsjusz_na_kelwin(T_c):
    if T_c < -273.15:
        return "Błąd: Temperatura poniżej zera absolutnego!"
    else:
        T_k = T_c + 273.15
        return T_k

wynik = celsjusz_na_kelwin(25)
print("Temperatura to:", wynik, "K")
```

## Argumenty domyślne (np. stałe fizyczne)

```python
def sila_grawitacji(m1, m2, r, G=6.674e-11):
    F = G * (m1 * m2) / r**2
    return F

# Wywołanie ze standardową stałą G
print(f"Siła: {sila_grawitacji(5.97e24, 7.34e22, 3.84e8):.2e} N")

# Możemy nadpisać G, podając je ręcznie:
# sila_grawitacji(m1, m2, r, G=1.0)
```
