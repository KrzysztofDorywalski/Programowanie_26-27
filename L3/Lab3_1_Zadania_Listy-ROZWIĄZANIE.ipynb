### Zadanie 1: Manipulacja listą
Mamy listę długości fali światła widzialnego. Zrób po kolei następujące kroki:
1. Dodaj na końcu listy wartość `750` używając `.append()`.
2. Wstaw na samym początku listy (indeks 0) wartość `380` używając `.insert()`.
3. Posortuj listę rosnąco i wyświetl ją.

```python
dlugosci_fali = [450, 600, 520, 700, 480]

# Twój kod:
# 1. Dodaj na końcu listy wartość 750 używając .append()
dlugosci_fali.append(750)

# 2. Wstaw na samym początku listy (indeks 0) wartość 380 używając .insert()
dlugosci_fali.insert(0, 380)

# 3. Posortuj listę rosnąco i wyświetl ją
dlugosci_fali.sort()
print(dlugosci_fali)
```

### Zadanie 2: Analiza zanieczyszczonych pomiarów
Masz podaną listę pomiarów, z których dwa skrajne (największy i najmniejszy) wiesz, że są błędne (szum aparatury). 
1. Użyj `.remove(wartość)`, aby usunąć wartość maksymalną i minimalną (wykorzystaj funkcje `max()` i `min()` aby je znaleźć).
2. Ze zmodyfikowanej listy oblicz nową średnią za pomocą `sum()` i `len()`.
3. Wypisz nową średnią.

```python
pomiary = [10.2, 10.5, 9.8, -50.0, 10.1, 10.8, 100.5, 10.3]

# Twój kod:
print(f"Lista początkowa: {pomiary}")

# 1. Usunięcie wartości maksymalnej i minimalnej za pomocą .remove(), max() i min()
max_wartosc = max(pomiary)
min_wartosc = min(pomiary)

pomiary.remove(max_wartosc)
pomiary.remove(min_wartosc)

print(f"Lista po usunięciu szumów ({min_wartosc} oraz {max_wartosc}): {pomiary}")

# 2. Obliczenie nowej średniej za pomocą sum() i len()
nowa_srednia = sum(pomiary) / len(pomiary)

# 3. Wypisanie nowej średniej
print(f"Nowa średnia pomiarów: {nowa_srednia}")
```

### Zadanie 3: Iteracja po liście i liczby pseudolosowe
Napisz skrypt, który za pomocą instrukcji for znajduje największą i najmniejszą liczbę ze zbioru n wylosowanych liczb całkowitych, generowanych losowo w przedziale od 0 do 100 (w zadaniu n = 5) oraz oblicza wartość średnią ze wszystkich wylosowanych liczb.

Porównaj obliczona wartość średniej dla, kolejno n = 50, 500, 5000

**Wskazówka: do losowania liczb pseudolosowych użyj  modułu random**

import random  #dołączenie modułu

random.randint(0, 100)

```python
#Twój kod
import random

# Część 1: Szczegółowa analiza dla n = 5
n = 5
wylosowane = []

# Generowanie n liczb losowych z przedziału [0, 100] przy użyciu pętli for
for _ in range(n):
    wylosowane.append(random.randint(0, 100))

# Inicjalizacja zmiennych do wyszukiwania ekstremów i sumy pierwszego elementu
maksimum = wylosowane[0]
minimum = wylosowane[0]
suma = 0

# Przetwarzanie wylosowanej listy za pomocą pętli for
for liczba in wylosowane:
    if liczba > maksimum:
        maksimum = liczba
    if liczba < minimum:
        minimum = liczba
    suma += liczba

srednia = suma / n

print(f"--- Wyniki dla n = {n} ---")
print(f"Wylosowane liczby: {wylosowane}")
print(f"Największa liczba: {maksimum}")
print(f"Najmniejsza liczba: {minimum}")
print(f"Średnia arytmetyczna: {srednia}")

print("\n--- Porównanie średniej dla różnych wartości n ---")
print("Wskazówka: Wraz ze wzrostem n średnia zbliża się do wartości oczekiwanej (50).")

# Część 2: Porównanie dla n = 5, 50, 500, 5000
for n in [5, 50, 500, 5000]:
    suma = 0
    for _ in range(n):
        suma += random.randint(0, 100)
    
    srednia = suma / n
    print(f"n = {n:5d} -> Średnia: {srednia:.2f}")
```

### Zadanie 4: Reguła dziel i zwyciężaj
Napisz skrypt – grę, w którym celem jest odgadnięcie wylosowanej liczby w zakresie od 1 do 100. Jeśli podana liczba jest większa od wylosowanej, program powinien o tym poinformować. Podobnie w przypadku, gdy podana liczba jest mniejsza od wylosowanej

```python
import random

print("Witaj w grze 'Zgadnij liczbę'!")
print("Wylosowałem dla Ciebie liczbę z zakresu od 1 do 100. Spróbuj ją odgadnąć.")

# Losowanie liczby od 1 do 100
wylosowana_liczba = random.randint(1, 100)
liczba_prob = 0

# Główna pętla gry
while True:
    try:
        # Pobranie próby od gracza
        proba = int(input("Podaj swoją propozycję: "))
        liczba_prob += 1
        
        # Sprawdzenie warunków
        if proba > wylosowana_liczba:
            print("Za dużo! Spróbuj mniejszej liczby.")
        elif proba < wylosowana_liczba:
            print("Za mało! Spróbuj większej liczby.")
        else:
            print(f"Gratulacje! Odgadłeś liczbę {wylosowana_liczba} w {liczba_prob} próbach.")
            break
            
    except ValueError:
        print("To nie jest poprawna liczba całkowita. Spróbuj ponownie.")
```
