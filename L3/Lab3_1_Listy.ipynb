## Lab 3: Listy

Lista to zbiór elementów, do których mamy dostęp przez ich numer (indeks). Zawsze liczymy od zera! 

```python
pomiary_czasu = [1.2, 1.4, 1.3, 1.5, 1.2]

print("Pierwszy pomiar:", pomiary_czasu[0])
print("Ostatni pomiar:", pomiary_czasu[-1])
print("Liczba wszystkich pomiarów (len):", len(pomiary_czasu))
```

## 1.1 Metody modyfikujące listę (append, insert, pop, remove)

```python
temperatury = [15.2, 16.5, 14.8, 17.1, 15.9]
print("Początkowa:", temperatury)

# .append(element) - dodaje na koniec
temperatury.append(18.0)
print("Po append:", temperatury)

# .insert(indeks, element) - wstawia w konkretne miejsce
temperatury.insert(1, 15.5) 
print("Po insert(1, 15.5):", temperatury)

# .remove(element) - usuwa pierwsze wystąpienie danej wartości
temperatury.remove(14.8)
print("Po remove(14.8):", temperatury)

# .pop() - usuwa OSTATNI element i zwraca jego wartość
ostatni = temperatury.pop()
print(f"Usunięty pop(): {ostatni}. Zostało: {temperatury}")
```

## 1.2 Wbudowane funkcje matematyczne (min, max, sum)
Nie musimy pisać pętli, aby znaleźć ekstrema lub sumę elementów listy.

```python
wyniki_eksperymentu = [2.3, 2.5, 2.1, 2.8, 2.4, 2.6]

print("Najmniejsza wartość:", min(wyniki_eksperymentu))
print("Największa wartość:", max(wyniki_eksperymentu))
print("Suma wartości:", sum(wyniki_eksperymentu))

# Obliczanie średniej w jednej linijce
srednia = sum(wyniki_eksperymentu) / len(wyniki_eksperymentu)
print(f"Średnia: {srednia:.2f}")
```

## 1.3 Sortowanie list (.sort() vs sorted())

```python
czasy_biegu = [12.5, 11.2, 13.1, 10.9, 11.8]

# .sort() modyfikuje oryginalną listę (rosnąco)
czasy_biegu.sort()
print("Posortowane rosnąco (.sort):", czasy_biegu)

# sorted() tworzy nową listę, zostawiając oryginał w spokoju
oryginalne = [5, 2, 8, 1]
nowe = sorted(oryginalne, reverse=True) # reverse=True dla malejącego
print("Oryginalna:", oryginalne)
print("Nowa malejąca (sorted):", nowe)
```

## 1.4 Iterowanie (pętla for) i przeszukiwanie (operator in)

```python
pomiary_katow = [45, 90, 180, 270]

if 90 in pomiary_katow:
    print("Znaleziono kąt 90 stopni!")

print("\nLista kątów z poprawką:")
for kat in pomiary_katow:
    kat_poprawiony = kat + 2.5
    print(f"Oryginał: {kat}, Po korekcie: {kat_poprawiony}")
```
