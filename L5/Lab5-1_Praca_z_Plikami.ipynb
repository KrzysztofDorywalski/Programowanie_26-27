### 1. Podstawowe otwarcie i zapis do pliku (złe praktyki)
Funkcja `open(nazwa_pliku, tryb)` pozwala otworzyć plik. Najważniejsze tryby to:
* `'r'` (read) - odczyt (domyślny)
* `'w'` (write) - zapis (kasuje zawartość pliku, jeśli istniał!)
* `'a'` (append) - dopisywanie na końcu pliku

```python
# Tworzymy plik z przykładowymi wynikami
plik = open("pomiary.txt", "w", encoding="utf-8")
plik.write("12.5\n")
plik.write("13.1\n")
plik.write("11.9\n")
plik.close() # BARDZO WAŻNE! Jeśli tego zapomnisz, dane mogą się nie zapisać, a plik zostanie zablokowany.
```

Dlaczego to zła praktyka? Bo jeśli program natrafi na błąd gdzieś w trakcie pracy, linijka `plik.close()` nigdy się nie wykona!

### 2. Prawidłowa praca z plikami: konstrukcja 'with'
Konstrukcja `with open(...) as ...:` (tzw. menedżer kontekstu) to bezpieczny standard. **Plik zamknie się automatycznie**, gdy tylko Python wyjdzie z bloku wcięć – nawet jeśli po drodze wystąpi błąd.

```python
# Odczytywanie całego pliku naraz:
with open("pomiary.txt", "r", encoding="utf-8") as f:
    zawartosc = f.read()
    print("Cała zawartość:\n" + zawartosc)

# Plik jest tu już zamknięty!
```

### 3. Czytanie pliku linia po linii (Iteracja)
Zazwyczaj z plików odczytujemy długie serie danych, więc najlepiej czytać je linijka po linijce za pomocą pętli `for`. Funkcja `.strip()` przydaje się, by usunąć niewidzialny znak nowej linii (`\n`).

```python
wyniki = []

with open("pomiary.txt", "r", encoding="utf-8") as f:
    for linia in f:
        czysta_linia = linia.strip()          # usuwa '\n' z końca
        liczba = float(czysta_linia)          # tekst na liczbę zmiennoprzecinkową
        wyniki.append(liczba)

print("Lista przetworzonych liczb:", wyniki)
print("Suma:", sum(wyniki))
```

### 4. Formatowanie i zapis raportu końcowego
Użyjemy trybu dopisywania (`'a'`), aby dołączyć wynik analizy do pliku bez kasowania surowych pomiarów.

```python
srednia = sum(wyniki) / len(wyniki)

with open("pomiary.txt", "a", encoding="utf-8") as f:
    f.write("------\n")
    f.write(f"Średnia: {srednia:.2f}\n")
    
print("Raport dopisany pomyślnie!")
```

### 5. Odczyt danych wielokolumnowych

**Uwaga:** skrypt poniżej zakłada istnienie danych pomiarowych zapisanych w pliku w trzech kolumnach

```python
czasy = []
polozenia = []
predkosci = []

with open("pomiary.txt", "r", encoding="utf-8") as plik:
    for linia in plik:
        linia = linia.strip()
        # Pomijamy linie komentarza i puste
        if not linia or linia.startswith("#"):
            continue
            
        # .split() bez argumentu dzieli tekst po dowolnym białym znaku (spacje, tabulatory)
        kolumny = linia.split()
        
        # Konwersja poszczególnych kolumn z napisu (str) na liczbę (float)
        czasy.append(float(kolumny[0]))
        polozenia.append(float(kolumny[1]))
        predkosci.append(float(kolumny[2]))

print("Wczytano pomiarów:", len(czasy))
```
