## Lab 5: Wprowadzenie do NumPy (Numerical Python)

Wizualizacja i obliczenia numeryczne w fizyce wymagają operowania na dużych zbiorach danych. Standardowe listy w Pythonie są elastyczne, ale wolne przy skomplikowanych obliczeniach. **NumPy** to fundamentalna biblioteka Pythona do obliczeń naukowych, która wprowadza szybkie i wydajne tablice jednorodne (`ndarray`).

### 1. Importowanie biblioteki i tworzenie tablic

Zgodnie z konwencją bibliotekę NumPy importujemy pod skróconą nazwą `np`.

```python
import numpy as np

# Tworzenie tablicy z istniejącej listy
pomiary_list = [1.25, 1.30, 1.28, 1.32]
pomiary_arr = np.array(pomiary_list)

print("Tablica NumPy:", pomiary_arr)
print("Typ obiektu:", type(pomiary_arr))
print("Typ danych wewnątrz tablicy:", pomiary_arr.dtype)
```

#### Dedykowane generatory tablic
W fizyce niezwykle często tworzymy osie czasu, siatki przestrzenne lub wektory wypełnione zerami/jedynkami.

```python
# 1. np.arange(start, stop, krok) - podobnie jak range(), ale obsługuje ułamki
t_krok = np.arange(0, 2.0, 0.4)
print("np.arange:", t_krok)

# 2. np.linspace(start, stop, liczba_punktów) - idealne do dyskretyzacji dziedziny funkcji
t_lin = np.linspace(0, 10, 5)
print("np.linspace:", t_lin)

# 3. Tablice zer i jedynek
zera = np.zeros(4)
jedynki = np.ones(4)
print("Zeros:", zera)
print("Ones:", jedynki)
```

### 2. Wektoryzacja – obliczenia bez pętli `for`!

Tradycyjna lista w Pythonie przelicza każdy element osobno w pętli. NumPy stosuje **wektoryzację**, wykonując operacje na całej tablicy naraz w zoptymalizowanym kodzie C.

```python
# Różnica w zachowaniu operatorów:
lista = [1, 2, 3]
tablica = np.array([1, 2, 3])

print("Lista * 2 (powielenie):", lista * 2)
print("Tablica * 2 (mnożenie elementów):", tablica * 2)

# Obliczenia fizyczne na wektorze:
# Wzór na drogę w ruchu jednostajnie przyspieszonym: s = 0.5 * a * t^2
a = 9.81  # m/s^2
t = np.linspace(0, 5, 6)  # czas od 0 do 5 s (6 punktów)

s = 0.5 * a * t**2

for czas, droga in zip(t, s):
    print(f"t = {czas:3.1f} s  -->  s = {droga:6.2f} m")
```

### 3. Funkcje uniwersalne (ufuncs) i trygonometria

NumPy dostarcza zoptymalizowane odpowiedniki funkcji matematycznych (`np.sin`, `np.cos`, `np.exp`, `np.sqrt` itd.), które przyjmują całe tablice.

```python
# Generujemy kąty od 0 do 2*pi
katy = np.linspace(0, 2 * np.pi, 5)

sin_vals = np.sin(katy)
cos_vals = np.cos(katy)

print("Kąty [rad]:", np.round(katy, 2))
print("Sinus:       ", np.round(sin_vals, 2))
print("Tożsamość (sin^2 + cos^2):", np.round(sin_vals**2 + cos_vals**2, 2))
```

### 4. Indeksowanie warunkowe (maski logiczne)

Zamiast pisać pętlę `for` i instrukcję `if` do odsiewania błędnych pomiarów, w NumPy tworzymy **maskę booleanową**.

```python
# Odczyty napięcia z czujnika z błędnymi wartościami ujemnymi oraz przepięciem
odczyty = np.array([3.3, 3.4, -99.0, 3.2, 105.0, 3.5, -0.1, 3.3])

# Maska logiczna dla zakresu fizycznego [0V, 5V]
maska = (odczyty >= 0) & (odczyty <= 5)

print("Maska wartości poprawnych:", maska)
print("Oczyszczone dane:        ", odczyty[maska])
```

### 5. Podstawowa statystyka zoptymalizowana w NumPy

Tablice NumPy posiadają wbudowane szybkie metody statystyczne.

```python
dane = odczyty[maska] # używamy oczyszczonych danych

print(f"Liczba pomiarów:        {dane.size}")
print(f"Średnia arytmetyczna:  {dane.mean():.3f} V")
print(f"Odchylenie standardowe: {dane.std():.3f} V")
print(f"Minimum i Maksimum:     {dane.min():.2f} V / {dane.max():.2f} V")
```

### 6. Odczyt pliku w NumPy

```python
# skiprows=1 pomija 1. linię nagłówkową
# unpack=True przypisuje kolumny bezpośrednio do zmiennych
czas, polozenie, predkosc = np.loadtxt("pomiary.txt", skiprows=1, unpack=True)

print("Czas:", czas)
print("Polożenie:", polozenie)

#czas, polozenie, predkosc = np.loadtxt("pomiary.csv", delimiter=",", skiprows=1, unpack=True)
```
