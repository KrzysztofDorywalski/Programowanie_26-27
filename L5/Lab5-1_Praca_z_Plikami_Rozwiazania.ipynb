## Lab 5: Operacje wejścia/wyjścia (I/O) - Praca z plikami tekstowymi

### Przygotowanie środowiska
Tworzenie pliku testowego `czujnik_lux.txt` z przykładowymi danymi i błędami.

```python
with open("czujnik_lux.txt", "w", encoding="utf-8") as f:
    f.write("850.4\n")
    f.write("ERROR\n")
    f.write("852.1\n")
    f.write("855.9\n")
    f.write("NO_DATA\n")
    f.write("849.0\n")
print("Plik czujnik_lux.txt został utworzony.")
```

### Zadanie 1: Kopiowanie z filtrowaniem (przetwarzanie "brudnych" danych)
Otwieramy plik źródłowy do odczytu i nowy plik do zapisu, przepisując tylko poprawne linie.

```python
with open("czujnik_lux.txt", "r", encoding="utf-8") as f_in:
    with open("czyste_lux.txt", "w", encoding="utf-8") as f_out:
        for linia in f_in:
            if "ERROR" not in linia and "NO_DATA" not in linia:
                f_out.write(linia)

print("Plik czyste_lux.txt został utworzony i przefiltrowany.")
```

### Zadanie 2: Analiza z użyciem funkcji i list
Funkcja `analizuj_plik` wczytuje liczby z pliku, wyznacza min, max oraz średnią i wyświetla sformatowany raport.

```python
def analizuj_plik(nazwa_pliku):
    pomiary = []
    
    with open(nazwa_pliku, "r", encoding="utf-8") as f:
        for linia in f:
            czysta = linia.strip()
            if czysta:
                pomiary.append(float(czysta))
    
    if not pomiary:
        print("Brak danych do analizy.")
        return
    
    minimum = min(pomiary)
    maksimum = max(pomiary)
    srednia = sum(pomiary) / len(pomiary)
    
    print(f"--- RAPORT DLA PLIKU: {nazwa_pliku} ---")
    print(f"Liczba poprawnych pomiarów: {len(pomiary)}")
    print(f"Najmniejsze natężenie: {minimum:.1f} lx")
    print(f"Największe natężenie:  {maksimum:.1f} lx")
    print(f"Średnie natężenie:     {srednia:.2f} lx")

# Przetestowanie funkcji
analizuj_plik("czyste_lux.txt")
```

### Zadanie z gwiazdką: Generator bazy danych CSV
Automatyczne wygenerowanie 100 wierszy w pliku CSV za pomocą pętli `for`.

```python
with open("moje_izotopy.csv", "w", encoding="utf-8") as f:
    for i in range(1, 101):
        f.write(f"{i}, Izotop_{i}\n")

print("Plik moje_izotopy.csv został wygenerowany.")
```
