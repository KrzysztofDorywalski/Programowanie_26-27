## Krok 0: Przygotowanie pliku pomiarowego
Uruchom poniższy kod, aby utworzyć plik `eksperyment_szum.csv` z przykładowymi danymi pomiarowymi na swoim dysku.

```python
import numpy as np

# Kod generujący plik ze zniekształconymi danymi eksperymentalnymi
np.random.seed(42)
with open("eksperyment_szum.csv", "w", encoding="utf-8") as f:
    f.write("Czas_s, Wychylenie_cm\n")
    t_raw = np.linspace(0, 10, 100)
    x_raw = 5.0 * np.exp(-0.25 * t_raw) * np.cos(2 * np.pi * t_raw) + np.random.normal(0, 0.15, 100)
    
    for i in range(len(t_raw)):
        if i == 12:
            f.write("1.21, ERR\n")
        elif i == 45:
            f.write("-99.0, 2.15\n")
        elif i == 78:
            f.write("7.88, 999.0\n")
        else:
            f.write(f"{t_raw[i]:.3f}, {x_raw[i]:.3f}\n")

print("Plik eksperyment_szum.csv został pomyślnie wygenerowany.")
```
