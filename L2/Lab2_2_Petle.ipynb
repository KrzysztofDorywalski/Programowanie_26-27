# Zajęcia 2 - Część 2: Pętle (for i while)

Pętle pozwalają na automatyzację powtarzalnych obliczeń. Pętla `for` jest idealna, gdy znamy z góry liczbę powtórzeń (np. chcemy wygenerować tabelę dla konkretnych sekund).

## Pętla for i range()
Obliczmy drogę w spadku swobodnym $s = \frac{gt^2}{2}$ dla pierwszych 5 sekund.

```python
g = 9.81
print("Czas [s] | Droga [m]")
print("--------------------")

# range(0, 6) wygeneruje liczby: 0, 1, 2, 3, 4, 5
for t in range(0, 6):
    s = 0.5 * g * t**2
    print(f"{t:8} | {s:.2f}")
```

## Pętla while
Pętla `while` wykonuje się, dopóki dany warunek fizyczny jest spełniony. Poniżej symulacja rozpadu promieniotwórczego.

```python
N = 1000        # początkowa liczba jąder
N_koncowe = 10  # szukana granica
T_pol = 5       # czas połowicznego rozpadu w dniach
dni = 0

while N > N_koncowe:
    N = N / 2
    dni += T_pol
    print(f"Po {dni} dniach zostało {N} jąder.")
```
