# Praca z tekstem (Stringi) w zastosowaniach fizycznych

W analizie danych często musimy radzić sobie z danymi wejściowymi w postaci tekstu (np. odczyty z czujników, logi z aparatury). Python posiada bardzo potężne narzędzia do manipulacji takimi łańcuchami znaków (typ `str`).

## 1. Indeksowanie i wycinanie (Slicing)
Tekst to ciąg znaków. Możemy "wyciągać" pojedyncze znaki lub ich fragmenty. W Pythonie liczymy od zera!

```python
odczyt = "Temperatura: 25.4 C"

print("Pierwszy znak:", odczyt[0])       # 'T'
print("Ostatni znak:", odczyt[-1])      # 'C' (indeksy ujemne liczą od końca)
print("Fragment (slicing):", odczyt[13:15]) # Wyciąga znaki od indeksu 13 do 16 (bez 17!)
```

## 2. Oczyszczanie danych: strip() i replace()
Dane z urządzeń pomiarowych często mają niepotrzebne spacje, znaki nowej linii (`\n`) lub używają przecinków zamiast kropek dziesiętnych.

```python
surowy_wynik = "   9,81\n"
print(f"Przed czyszczeniem: '{surowy_wynik}'")

# strip() usuwa białe znaki (spacje, entery) z początku i końca
czysty_wynik = surowy_wynik.strip()

# replace(co, na_co) zamienia fragmenty tekstu
wynik_kropka = czysty_wynik.replace(",", ".")

print(f"Po czyszczeniu: '{wynik_kropka}'")

# Teraz możemy zamienić tekst na liczbę zmiennoprzecinkową (float)
przyspieszenie = float(wynik_kropka)
print(f"Obliczenia: {przyspieszenie * 2}")
```

## 3. Przeszukiwanie tekstu: in, find()
Sprawdzanie, czy dany fragment znajduje się w tekście.

```python
log_aparatu = "OSTRZEZENIE: Niskie ciśnienie w komorze próżniowej."

# Użycie operatora 'in' (zwraca True lub False)
if "OSTRZEZENIE" in log_aparatu:
    print("Znaleziono ostrzeżenie!")

# Metoda find() zwraca indeks, na którym zaczyna się szukany tekst (lub -1, gdy nie znajdzie)
indeks = log_aparatu.find("Niskie")
print(f"Słowo zaczyna się od indeksu: {indeks}")
```
