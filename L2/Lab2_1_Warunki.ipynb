## Instrukcje if, elif, else
Zobaczmy, jak zabezpieczyć nasze obliczenia fizyczne. Sprawdzimy, czy prędkość cząstki nie przekracza prędkości światła $c$.

```python
c = 3e8 # m/s (prędkość światła)
v = 3.5e8 # prędkość ciała z naszego eksperymentu

if v >= c:
    print("Błąd: Prędkość ciała nie może przekraczać prędkości światła!")
elif v > 0.1 * c:
    print("Uwaga: Należy zastosować poprawki relatywistyczne.")
else:
    print("Można stosować mechanikę klasyczną.")
```
