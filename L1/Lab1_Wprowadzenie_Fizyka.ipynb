## 4. Ćwczenia: Kalkulator fizyczny
### Zadanie 1: Energia kinetyczna i potencjalna
Obliczmy energię dla ciała o masie $m$, poruszającego się z prędkością $v$ na wysokości $h$.

```python
m = 2.5  # kg
v = 15.0 # m/s
h = 10.0 # m
g = 9.81 # m/s^2

E_k = 0.5 * m * v**2
E_p = m * g * h
E_c = E_k + E_p

print(f"Energia kinetyczna: {E_k} J")
print(f"Energia potencjalna: {E_p} J")
print(f"Energia całkowita: {E_c} J")
```

### Zadanie 2: Prawo Powszechnego Ciążenia
Poznajemy notację naukową dla dużych i małych liczb (`e`). Wzór to:
$$F = G \frac{M_1 M_2}{R^2}$$

```python
G = 6.674e-11
M_ziemi = 5.972e24
M_ksiezyca = 7.348e22
R = 3.844e8

F = G * (M_ziemi * M_ksiezyca) / R**2
print(f"Siła grawitacji: {F:.2e} N")
```

### Zadanie 3: Okres drgań wahadła (importowanie bibliotek)
Wzór to:
$$T = 2\pi\sqrt{\frac{l}{g}}$$
Do liczby pi oraz pierwiastka potrzebujemy zaimportować bibliotekę `math`.

```python
import math

l = 1.5 # długość w metrach
g = 9.81

T = 2 * math.pi * math.sqrt(l/g)
print(f"Okres drgań wahadła o długości {l} m to: {T:.2f} s")
```
