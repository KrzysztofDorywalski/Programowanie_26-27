### 1. Wprowadzenie
Funkcja rekurencyjna to taka funkcja, która w swoim ciele wywołuje **samą siebie**.
Aby rekurencja nie trwała w nieskończoność (co doprowadziłoby do błędu `RecursionError`), funkcja musi posiadać **warunek końcowy** (warunek bazowy), przy którym przestaje wywoływać samą siebie i zaczyna zwracać wyniki w górę.

```python
def wiadomosc(i):
    if i > 0:  # Warunek końcowy: jeśli i <= 0, rekurencja się zatrzyma
        print(f'To jest funkcja rekurencyjna. Zostało {i} wywołań.')
        wiadomosc(i - 1)  # Funkcja wywołuje samą siebie z mniejszym 'i'

wiadomosc(5)
```

### 2. Przykład: silnia
Silnia z liczby $n$ (oznaczana jako $n!$) to iloczyn wszystkich liczb naturalnych od $1$ do $n$.
Możemy zapisać to iteracyjnie (używając pętli) lub rekurencyjnie (używając funkcji wywołującej samą siebie).

```python
# Rozwiązanie rekurencyjne
def silnia_rek(n):
    if n > 1:
        return n * silnia_rek(n - 1) # np. dla n=5: 5 * silnia_rek(4)
    else:
        return 1

# Rozwiązanie iteracyjne
def silnia_iter(n):
    silnia_temp = 1
    if n > 1:
        for i in range(2, n + 1):
            silnia_temp = silnia_temp * i
        return silnia_temp
    else:
        return 1

print("Silnia rekurencyjna (5!):", silnia_rek(5))
print("Silnia iteracyjna (5!):", silnia_iter(5))
```

### 3. Pomiar wydajności (moduł timeit)
Sprawdźmy, które podejście jest szybsze w Pythonie: rekurencja czy pętla `for`?

```python
import timeit

# Mierzymy czas dla rekurencji
kod_rek = """
def silnia_rek(n):
    if n > 1:
        return n * silnia_rek(n - 1)
    else:
        return 1
silnia_rek(20)
"""
czas_rek = timeit.timeit(kod_rek, number=100000) # 100 tysięcy wywołań

# Mierzymy czas dla iteracji
kod_iter = """
def silnia_iter(n):
    silnia_temp = 1
    if n > 1:
        for i in range(2, n + 1):
            silnia_temp = silnia_temp * i
        return silnia_temp
    else:
        return 1
silnia_iter(20)
"""
czas_iter = timeit.timeit(kod_iter, number=100000)

print(f"Czas rekurencji: {czas_rek:.4f} s")
print(f"Czas iteracji:   {czas_iter:.4f} s")
```
