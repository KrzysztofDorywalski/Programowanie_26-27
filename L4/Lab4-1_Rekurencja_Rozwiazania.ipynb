#### Zadanie 1: Ciąg Fibonacciego
Rozwiązanie wykorzystujące dwa wywołania rekurencyjne w jednej linii.

```python
def fib_rek(n):
    # Warunki brzegowe
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Krok rekurencyjny
    else:
        return fib_rek(n-1) + fib_rek(n-2)

# Testowanie funkcji dla pierwszych 10 wyrazów
print("Pierwsze 10 liczb ciągu Fibonacciego:")
for i in range(10):
    print(f"fib({i}) = {fib_rek(i)}")
```

#### Zadanie 2: Rekurencyjne zdefiniowanie ciągu (wzór)
Zastosowanie operatora modulo `%` do weryfikacji parzystości i wykonania odpowiedniego kroku rekurencyjnego.

```python
def ciag_wzor(n):
    # Warunek bazowy
    if n == 0:
        return 1
    # Krok dla n parzystych
    elif n % 2 == 0:
        return (ciag_wzor(n-1))**2 - 1
    # Krok dla n nieparzystych
    else:
        return ciag_wzor(n-1) * 2 + 1

# Obliczenie kilku pierwszych wartości
print("Wartości ciągu:")
for i in range(6):
    print(f"a_{i} = {ciag_wzor(i)}")
```

#### Zadanie 3: Liczby trójkątne
Obliczanie i wypisywanie kolejnych liczb z użyciem rekurencji.

```python
def liczba_trojkatna(n):
    # Warunek bazowy
    if n == 1:
        return 1
    # Krok rekurencyjny
    else:
        return liczba_trojkatna(n-1) + n

print("Pierwsze 10 liczb trójkątnych:")
for i in range(1, 11): # Liczymy od 1 do 10 (bez 11)
    print(f"T_{i} = {liczba_trojkatna(i)}")
```
