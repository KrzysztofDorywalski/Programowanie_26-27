## Lab4: Struktury danych w Pythonie

Oprócz list, Python oferuje inne wbudowane struktury, które są niezwykle przydatne w pracy z danymi naukowymi: Krotki, Zbiory i Słowniki.

#### 1. Krotki (Tuples) - niezmienne struktury
Krotka to struktura używająca nawiasów okrągłych `()`. W przeciwieństwie do listy, po jej utworzeniu nie można dodawać, usuwać ani zmieniać jej elementów. Używamy ich do danych, które nie powinny ulec zmianie.

```python
# Definiowanie krotki
wspolrzedne_wektora = (10.5, 3.2, -1.0)

# Odczytywanie działa tak samo jak w listach
print("Współrzędna Y:", wspolrzedne_wektora[1])

# Próba modyfikacji wywoła błąd! (Odkoduj linijkę poniżej, aby sprawdzić)
# wspolrzedne_wektora[0] = 5.0
```

#### 2. Zbiory (Sets) - Unikalność i operacje logiczne
Zbiory przechowują nieuporządkowane kolekcje **unikalnych** elementów. Używają nawiasów klamrowych `{}`. Są idealne do szybkiego usuwania duplikatów z list.

```python
# Szybkie usunięcie duplikatów z listy
pomiary_z_bledami = [1.2, 1.5, 1.2, 1.8, 1.5, 1.2]
unikalne_pomiary = list(set(pomiary_z_bledami))
print("Unikalne pomiary:", unikalne_pomiary)

# Operacje matematyczne na zbiorach
detektor_a = {"alfa", "beta", "gamma"}
detektor_b = {"neutron", "beta", "gamma"}

print("Wykryte przez OBA detektory (przecięcie &):", detektor_a & detektor_b)
print("Wykryte TYLKO przez detektor A (różnica -):", detektor_a - detektor_b)
print("Wszystkie wykryte cząstki (suma |):", detektor_a | detektor_b)
```

#### 3. Słowniki (Dictionaries) - Mapowanie klucz-wartość
Słowniki przechowują dane pod konkretnymi nazwami (kluczami), a nie indeksami. Używają nawiasów klamrowych `{}` i dwukropka `:`.

```python
# Definiowanie stałych fizycznych
stale_fizyczne = {
    "c": 299792458,       # prędkość światła
    "G": 6.674e-11,       # stała grawitacyjna
    "h": 6.626e-34        # stała Plancka
}

# Pobieranie danych po kluczu
print(f"Prędkość światła to: {stale_fizyczne['c']} m/s")

# Bezpieczne pobieranie (.get)
# Szukamy stałej Boltzmanna, której nie ma w słowniku. Zamiast błędu dostajemy "Brak danych"
print("Stała Boltzmanna:", stale_fizyczne.get("k", "Brak danych"))

# Dodanie nowego elementu lub modyfikacja
stale_fizyczne["g"] = 9.81
print("Dodano g:", stale_fizyczne["g"])
```
